## Part A — Supervised learning

We will use the **Iris** dataset built into scikit-learn. Every flower example has four measurements and a known species label.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X, y = iris.data, iris.target
print("Samples:", X.shape[0])
print("Features:", iris.feature_names)
print("Classes:", list(iris.target_names))
```

### Split before fitting

The classifier learns from the **training split**. We measure accuracy on a separate **test split** that it did not see during training. `stratify=y` keeps the class proportions similar.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print(f"Accuracy on held-out examples: {accuracy_score(y_test, predictions):.3f}")
print("True classes:     ", y_test[:10].tolist())
print("Predicted classes:", predictions[:10].tolist())
```

**Think:** What is the difference between `X` (input measurements) and `y` (target labels)? What does accuracy tell us, and what does it *not* tell us?

## Part B — Unsupervised learning

Now pretend the labels are missing. **K-means** groups flowers using only measurements. Cluster IDs are arbitrary numbers, not necessarily flower species.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

X_petals = iris.data[:, 2:4]  # Petal length and width only
clusters = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(X_petals)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].scatter(X_petals[:, 0], X_petals[:, 1], c=clusters, cmap="viridis", s=30)
axes[0].set_title("K-means clusters (unlabeled learning)")
axes[1].scatter(X_petals[:, 0], X_petals[:, 1], c=iris.target, cmap="viridis", s=30)
axes[1].set_title("Actual species (shown for comparison)")
for ax in axes:
    ax.set_xlabel("Petal length (cm)")
    ax.set_ylabel("Petal width (cm)")
plt.tight_layout()
plt.show()
```
